```python
import pandas as pd
import random

# Sanctioned countries excluded
EXCLUDED_COUNTRIES = {"RUS", "BLR", "IRI", "SYR", "MYA", "SUD", "PRK"}

# Available countries from your dataset
COUNTRIES = [
    "CHI", "ARG", "COL", "HUN", "GER", "USA", "CAN", "MEX", "ITA", "JPN",
    "NED", "BEL", "FRA", "GBR", "CZE", "SRB", "TPE", "NZL", "AUS", "SGP",
    "HKG", "CHN", "POL", "LUX", "SUI", "ROU", "CRC", "NGR", "ESP", "PAR",
    "BRA", "ISR", "AUT", "SWE", "UKR", "SVK", "MDA", "UAE", "DEN", "THA",
    "MAS", "BUL", "GRE", "ALB", "SMR", "PER", "POR", "CMR", "IRL", "LAT",
    "ECU", "GUA", "IND", "ALG", "LIE", "CRO", "URU", "NOR", "MAR", "VIE",
    "VEN", "RSA", "GAB", "TUR", "LTU", "FIN", "TUN", "QAT", "CYP", "ZIM",
    "MOZ", "LES", "MAW", "BOT", "NAM", "SWZ", "ZAM", "MRI", "KOR", "SRI",
    "HON", "BOL", "PUR", "UZB", "BIH", "MNE", "MKD", "EST", "JOR", "UGA",
    "NEP", "EGY", "KEN", "JAM", "ESA", "MLT", "AND", "DOM", "ANG", "KOS",
    "ISL", "ETH", "OMA", "KAZ", "AZE", "FIJ", "GEO", "KUW", "GHA", "PAK"
]

COUNTRIES = [c for c in COUNTRIES if c not in EXCLUDED_COUNTRIES]

random.seed(42)

def generate_performance():
    hours = random.randint(8, 20)
    minutes = random.randint(0, 59)
    seconds = random.randint(0, 59)
    return f"{hours:02d}:{minutes:02d}:{seconds:02d}"

def generate_avg_speed(performance):
    parts = performance.split(":")
    total_hours = int(parts[0]) + int(parts[1]) / 60 + int(parts[2]) / 3600
    return round(100 / total_hours, 2)

def generate_age_category(year_of_birth, gender):
    age = 2024 - year_of_birth
    brackets = [23, 35, 40, 45, 50, 55, 60, 65, 70]
    for b in brackets:
        if age < b:
            return f"{gender}{b}"
    return f"{gender}70"

rows = []
for athlete_id in range(90001, 90201):
    gender = random.choice(["M", "W"])
    year_of_birth = random.randint(1965, 2000)
    performance = generate_performance()
    avg_speed = generate_avg_speed(performance)
    country = random.choice(COUNTRIES)
    age_category = generate_age_category(year_of_birth, gender)

    rows.append({
        "Year of event": 2024,
        "Event dates": "15.06.2024",
        # Event name follows same format as original data: "Event Name (HOST_COUNTRY)"
        # WHY: ESP = Spain, where Marathos Iberia Ultra is hosted
        "Event name": "Marathos Iberia Ultra (ESP)",
        "Event distance/length": "100km",
        "Event number of finishers": 200,
        "Athlete performance": performance,
        "Athlete club": "",
        "Athlete country": country,
        "Athlete year of birth": year_of_birth,
        "Athlete gender": gender,
        "Athlete age category": age_category,
        "Athlete average speed": avg_speed,
        "Athlete ID": athlete_id
    })

df = pd.DataFrame(rows)

# Save to Unity Catalog volume
output_path = "/Volumes/marathos/default/raw/events/marathos_iberia_ultra_2024.csv"
df.to_csv(output_path, index=False)

print(f"✅ Saved {len(df)} athletes to {output_path}")
display(df.head(10))
```
